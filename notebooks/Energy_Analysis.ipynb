```python
import pandas as pd

df = pd.read_csv("../data/raw/Steel_industry_data.csv")

print("Dataset Loaded Successfully!")
```

```python
df.info()
```

```python
df.describe()
```

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
df.dtypes
```

```python
df['date'] = pd.to_datetime(df['date'])
```

```python
df['date'].head(10)
```

```python
df['date'] = pd.to_datetime(
    df['date'],
    format='%d/%m/%Y %H:%M'
)
```

```python
df.dtypes
```

```python
df['Year'] = df['date'].dt.year
```

```python
df['Month'] = df['date'].dt.month_name()
```

```python
df['Day'] = df['date'].dt.day
```

```python
df['Hour'] = df['date'].dt.hour
```

```python
df.head()
```

```python
df['Hour'] = df['date'].dt.hour
df['Time'] = df['date'].dt.strftime('%H:%M')
```

```python
df.head()
```

```python
df['Usage_kWh'].describe()
```

```python
plt.figure(figsize=(8,5))

plt.hist(df['Usage_kWh'], bins=30)

plt.title('Distribution of Energy Consumption')
plt.xlabel('Energy Consumption (kWh)')
plt.ylabel('Frequency')

plt.show()
```

```python
df.groupby('Load_Type')['Usage_kWh'].mean()
```

```python
load = df.groupby('Load_Type')['Usage_kWh'].mean()

plt.figure(figsize=(6,4))

load.plot(kind='bar')

plt.title('Average Energy Consumption by Load Type')
plt.ylabel('Average Usage (kWh)')

plt.show()
```

```python
df.groupby('WeekStatus')['Usage_kWh'].mean()
```

```python
week = df.groupby('WeekStatus')['Usage_kWh'].mean()

plt.figure(figsize=(5,4))

week.plot(kind='bar')

plt.title('Weekday vs Weekend Energy Consumption')
plt.ylabel('Average Usage (kWh)')

plt.show()
```

```python
day = df.groupby('Day_of_week')['Usage_kWh'].mean()

day
```

```python
plt.figure(figsize=(8,5))

day.plot(kind='bar')

plt.title('Average Energy Consumption by Day')
plt.ylabel('Average Usage (kWh)')

plt.show()
```

```python
hour = df.groupby('Hour')['Usage_kWh'].mean()

hour
```

```python
plt.figure(figsize=(10,5))

hour.plot()

plt.title('Hourly Energy Consumption')
plt.xlabel('Hour')
plt.ylabel('Average Usage (kWh)')

plt.grid()

plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.scatter(df['Usage_kWh'], df['CO2(tCO2)'], alpha=0.5)

plt.title('Energy Consumption vs CO₂ Emissions')
plt.xlabel('Energy Consumption (kWh)')
plt.ylabel('CO₂ Emissions (tCO₂)')

plt.show()
```

```python
co2 = df.groupby('Load_Type')['CO2(tCO2)'].mean()

co2
```

```python
plt.figure(figsize=(6,4))

co2.plot(kind='bar')

plt.title('Average CO₂ Emissions by Load Type')
plt.ylabel('Average CO₂')

plt.show()
```

```python
df[['Lagging_Current_Power_Factor',
    'Leading_Current_Power_Factor']].describe()
```

```python
plt.figure(figsize=(8,5))

plt.hist(df['Lagging_Current_Power_Factor'], bins=25)

plt.title('Lagging Power Factor Distribution')

plt.xlabel('Power Factor')

plt.show()
```

```python
numeric_df = df.select_dtypes(include='number')

numeric_df.corr()
```

```python
import seaborn as sns

plt.figure(figsize=(10,8))

sns.heatmap(
    numeric_df.corr(),
    annot=True,
    cmap='coolwarm'
)

plt.title('Correlation Matrix')

plt.show()
```
